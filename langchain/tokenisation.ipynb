```python
from pypdf import PdfReader

reader = PdfReader("harrypotter.pdf")

print("Pages:", len(reader.pages))
text=""
```

```python
for page in reader.pages[0:50]:
    text += page.extract_text()
```

```python
print(text)
```

```python
print(len(text))
```

```python
#our goal is to tokenize all of the charaters in this text into individual tokens. later turn them into eembeddings and store them in a vector database.
```

```python
import re 

preprocessed = re.split(r'([,.:;?_!"()\'}|__|\s)])', text)
preprocessed = [item.strip() for item in preprocessed if item.strip()]
print(preprocessed[:100])

```

```python
print(len(preprocessed))
```

```python
#convert token into token ids
```

```python
all_words = sorted(set(preprocessed))
```

```python
print(len(all_words))
```

```python
vocab = {word : idx for idx,word in enumerate(all_words)}
```

```python
for item in vocab.items():
    print(item)
```
