### TOOLS
models can request to call tools that perform tasks such as fetching data from a database, searching the web or running code. tools are pairings of schema and a function to execute.

```python
import os 
from langchain import ChatGroq
model = ChatGroq(model_name="groq:chat-groq-1")
response = model.invoke("Hello, how are you?")
response
```

```python
from lomgchain.tools import tool

@tool
def get_weather(location:str) -> str:
    """Get the current weather for a given location."""
    # Here you would implement the logic to get the weather
    return f"The current weather in {location} is sunny with a temperature of 25°C."

model_with_tools=model.bind_tools([get_weather], model=model)
 
```

### tool execution loops


```python
messages = [
    {"role": "user", "content": "whats the weather in New York?"},
]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

#execute tools and collect results 
for tool_call in ai_msg.tool_calls:
    tool_name = tool_call.name
    tool_args = tool_call.args
    tool_result = model_with_tools.execute_tool(tool_name, **tool_args)
    messages.append({"role": "tool", "content": tool_result})
```
