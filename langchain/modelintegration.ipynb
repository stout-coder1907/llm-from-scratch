### streaming and batch 

most models can stream their output while it is being generated. By displaying ouptut progressively, streaming significantly improves user experience, particularly for longer responses 

```python
for chunk in models.stream("write me a 200 words paragraph on artificial intelligence and its impact on society")
    print(chunk.text, end="|", flush=True)
```

### batch by batch processing  
batching is the collection of independent requests to a model which can significanlty improve performance and reduce costs as the processing can be done in paralllel


```python
responses= models.batch([
    "write me a 200 words paragraph on artificial intelligence and its impact on society",
    "write me a 200 words paragraph on artificial intelligence and its impact on society",
    "write me a 200 words paragraph on artificial intelligence and its impact on society"],
    config={"max_cuurency": 1000, "temperature": 0.5, "top_p": 0.9})

for response in responses:
    print(response)
```
